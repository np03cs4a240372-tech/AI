# House Price Dataset Analysis
This notebook explains each step clearly and uses the provided CSV dataset.

```python
from google.colab import drive
drive.mount('/content/drive')
```

## 1. Import Libraries and Load Dataset

```python
import pandas as pd

df = pd.read_csv('/content/drive/MyDrive/ConceptAndTechnologiesOfAI/Houseprice (1).csv')
df.head()
```

## 2. Dataset Information
Shows column names, data types, and missing values.

```python
df.info()
```

## 3. Statistical Summary
Provides mean, median, min, max, and quartiles.

```python
df.describe()
```

## 4. Check Missing Values

```python
df.isnull().sum()
```

## 5. Handle Missing Values
Fill missing numerical values using median.

```python
df = df.fillna(df.median(numeric_only=True))
df.isnull().sum()
```

## 6. Correlation Matrix
Helps understand relationships between numerical variables.

```python
df.corr(numeric_only=True)
```

## 7. House Price Distribution

```python
import matplotlib.pyplot as plt

df['HousePrice'].hist()
plt.xlabel('Price')
plt.ylabel('Frequency')
plt.title('House Price Distribution')
plt.show()
```

## 8. Feature vs Price Relationship
Example using first feature column.

```python
feature = df.columns[0]

plt.scatter(df[feature], df['HousePrice'])
plt.xlabel(feature)
plt.ylabel('Price')
plt.title(f'{feature} vs Price')
plt.show()
```
