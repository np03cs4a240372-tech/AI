# Worksheet 6 – Implementation of Sigmoid Regression

## 1. Sigmoid (Logistic) Function

```python

import numpy as np

def logistic_function(x):
    return 1 / (1 + np.exp(-x))

# Test
def test_logistic_function():
    assert round(logistic_function(0),3) == 0.5
    assert round(logistic_function(2),3) == round(1/(1+np.exp(-2)),3)
    assert round(logistic_function(-3),3) == round(1/(1+np.exp(3)),3)
    arr = np.array([0,2,-3])
    expected = np.array([0.5,0.881,0.047])
    assert np.all(np.round(logistic_function(arr),3)==expected)
    print("Sigmoid tests passed")

test_logistic_function()
```

## 2. Log Loss Function

```python

def log_loss(y_true, y_pred):
    y_pred = np.clip(y_pred, 1e-10, 1-1e-10)
    return -(y_true*np.log(y_pred) + (1-y_true)*np.log(1-y_pred))

def test_log_loss():
    assert np.isclose(log_loss(1,1),0.0)
    assert np.isclose(log_loss(0,0),0.0)
    assert np.isclose(log_loss(1,0.8),-(np.log(0.8)))
    print("Log-loss tests passed")

test_log_loss()
```

## 3. Cost Function

```python

def cost_function(y_true, y_pred):
    n = len(y_true)
    loss_vec = [log_loss(y_true[i], y_pred[i]) for i in range(n)]
    return np.sum(loss_vec)/n

# Test
y_true = np.array([1,0,1])
y_pred = np.array([0.9,0.1,0.8])
print("Cost:", cost_function(y_true,y_pred))
```

## 4. Cost Function with Model Parameters

```python

def costfunction_logreg(X, y, w, b):
    z = np.dot(X,w) + b
    y_pred = logistic_function(z)
    return cost_function(y, y_pred)

X = np.array([[10,20],[-10,10]])
y = np.array([1,0])
w = np.array([0.5,1.5])
b = 1
print(costfunction_logreg(X,y,w,b))
```

## 5. Gradient Computation

```python

def compute_gradient(X, y, w, b):
    n = X.shape[0]
    z = np.dot(X,w) + b
    y_pred = logistic_function(z)
    grad_w = -(1/n)*np.dot(X.T,(y - y_pred))
    grad_b = -(1/n)*np.sum(y - y_pred)
    return grad_w, grad_b

grad_w, grad_b = compute_gradient(X,y,w,b)
print(grad_w, grad_b)
```

## 6. Gradient Descent

```python

def gradient_descent(X, y, w, b, alpha, n_iter):
    cost_history = []
    for _ in range(n_iter):
        grad_w, grad_b = compute_gradient(X,y,w,b)
        w -= alpha*grad_w
        b -= alpha*grad_b
        cost_history.append(costfunction_logreg(X,y,w,b))
    return w, b, cost_history

X = np.array([[0.1,0.2],[-0.1,0.1]])
y = np.array([1,0])
w = np.zeros(2)
b = 0
w,b,cost = gradient_descent(X,y,w,b,0.1,1000)
print(w,b,cost[-1])
```

## 7. Prediction Function

```python

def prediction(X, w, b, threshold=0.5):
    probs = logistic_function(np.dot(X,w)+b)
    return (probs >= threshold).astype(int)

X_test = np.array([[0.5,1.0],[1.5,-0.5],[-0.5,-1.0]])
w_test = np.array([1,-1])
b_test = 0
print(prediction(X_test,w_test,b_test))
```

## 8. Evaluation Metrics

```python

def evaluate_classification(y_true, y_pred):
    TP = np.sum((y_true==1)&(y_pred==1))
    TN = np.sum((y_true==0)&(y_pred==0))
    FP = np.sum((y_true==0)&(y_pred==1))
    FN = np.sum((y_true==1)&(y_pred==0))

    precision = TP/(TP+FP) if TP+FP>0 else 0
    recall = TP/(TP+FN) if TP+FN>0 else 0
    f1 = 2*precision*recall/(precision+recall) if precision+recall>0 else 0

    return {
        "confusion_matrix": np.array([[TN,FP],[FN,TP]]),
        "precision": precision,
        "recall": recall,
        "f1_score": f1
    }

print(evaluate_classification(np.array([1,0,1,0]),np.array([1,0,0,0])))
```
