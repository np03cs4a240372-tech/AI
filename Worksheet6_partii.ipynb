```python

import numpy as np
import pandas as pd
```

```python

# Load dataset
data = pd.read_csv('/content/drive/MyDrive/ConceptAndTechnologiesOfAI/Iris.csv')
data.head()
```

```python

# Drop Id column if present
if 'Id' in data.columns:
    data = data.drop('Id', axis=1)

print(data.columns)
```

```python

# Separate features and target
X = data.iloc[:, :-1].values
y = data.iloc[:, -1].values
```

```python

# Encode class labels
classes = np.unique(y)
class_map = {cls: i for i, cls in enumerate(classes)}
y_encoded = np.array([class_map[label] for label in y])
```

```python

# One-hot encoding
def one_hot(y, num_classes):
    encoded = np.zeros((len(y), num_classes))
    for i, val in enumerate(y):
        encoded[i, val] = 1
    return encoded

Y = one_hot(y_encoded, len(classes))
```

```python

# Train-test split
def train_test_split_manual(X, Y, test_size=0.2):
    split = int(len(X) * (1 - test_size))
    return X[:split], X[split:], Y[:split], Y[split:]

X_train, X_test, Y_train, Y_test = train_test_split_manual(X, Y)
```

```python

# Softmax function
def softmax(z):
    exp_z = np.exp(z - np.max(z, axis=1, keepdims=True))
    return exp_z / np.sum(exp_z, axis=1, keepdims=True)
```

```python

# Cross entropy loss
def cross_entropy(Y, Y_hat):
    return -np.mean(np.sum(Y * np.log(Y_hat + 1e-9), axis=1))
```

```python

# Training using gradient descent
def train_softmax(X, Y, lr=0.01, epochs=1000):
    W = np.zeros((X.shape[1], Y.shape[1]))
    loss_history = []

    for _ in range(epochs):
        logits = np.dot(X, W)
        Y_hat = softmax(logits)
        loss = cross_entropy(Y, Y_hat)
        loss_history.append(loss)
        grad = np.dot(X.T, (Y_hat - Y)) / len(X)
        W -= lr * grad

    return W, loss_history
```

```python

W, loss_history = train_softmax(X_train, Y_train)
```

```python

# Prediction
def predict(X, W):
    return np.argmax(softmax(np.dot(X, W)), axis=1)

y_pred = predict(X_test, W)
y_true = np.argmax(Y_test, axis=1)

accuracy = np.mean(y_pred == y_true)
print("Accuracy:", accuracy)
```
